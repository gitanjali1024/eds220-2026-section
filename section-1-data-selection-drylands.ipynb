## About the data:

This data contains occurrence records and associated climate and environmental data for all arthropod prey of listed species in California drylands. The data was not collected by the authors, rather it was compiled from GBIF data from 1945 to 2022, and does come with associated metadata. The data in this file is not sensitive. 

### citation: 
Rachel King, Jenna Braun, Michael Westphal, & CJ Lortie. (2023). Compiled occurrence records for prey items of listed species found in California drylands with associated environmental data. Knowledge Network for Biocomplexity. doi:10.5063/F1VM49RH.
